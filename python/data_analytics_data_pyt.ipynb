```python
# Remove duplicate network log entries
df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.info()
```

```python
df.columns = df.columns.str.replace(" ", "_")
```

```python
df.info()
```

```python
# Remove invalid or unrealistic bandwidth values
df = df[df["bandwidth"] > 0]


df = df[df["bandwidth"] < 5000]

```

```python
# Analyze average bandwidth usage across departments
department_traffic = (
    df.groupby("department")["bandwidth"]
    .mean()
    .sort_values(ascending=False)
)

print(department_traffic)
```

```python
# Filter suspicious traffic logs and analyze protocol threat frequency
threats = df[df["threat_flag"] == "YES"]

protocol_threats = threats["protocol"].value_counts()

print(protocol_threats)
```

```python
# Evaluate average packet loss across VLANs
vlan_performance = (
    df.groupby("vlan_id")["packet_loss_percent"]
    .mean()
    .sort_values(ascending=False)
)

print(vlan_performance)
```
